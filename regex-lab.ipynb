```python
# The finditer() method allows you to loop through the match string and do something for each return.
# Write a block of code that searches the contents of the mbox-short.txt file to identify IP addresses,
# then writes each to file called ip-addresses.txt. Add the output file to your repo with the notebook.

#get python to read the mbox file

import re

path = "C:\\Users\\erinr\\Downloads\\mbox-short.txt"
with open(path, 'r') as mbox:
    mbox_file = mbox.read()

# figure out the regex

# My thoughts: IP addresses are formatted as either IPv4,
# which are four sets of numbers separated by periods or IPv6 that are
# eight groups of letters and numbers separated by colons. Since this text file
# deals with emails, I will stick to finding IPv4 addresses.
#  I know that the numbers can only go up to 255, so 3 digits is the max.


# example: [127.0.0.1]

# I probably don't want to count repeats. I can also use compile() to define my regex.

# the numbers can only be 0-9 and there are at most 3 digits in each group
# [0-9]{1,3} means it will match 1 to 3 digits that are betewen 0 and 9.

# there are dots between the groups. I should use \. because just a period
# would represent any character.

# \b would also be useful for matching only the IP address and not any other numbers with periods in them. 

# It needs to start and end with \b and have four groups of [0-9]{1,3} separated by \. in between.

ip_address_finder = re.compile(r'\b[0-9]{1,3}\.[0-9]{1,3}\.[0-9]{1,3}\.[0-9]{1,3}\b')


# now for the python loop to write it to a text file after finding IP addresses
##### 'r' and 'w' mean read and write. I want to write a new file for the list.
##### dig back through SI506 notebook for how to format.

with open ("ip-addresses.txt", 'w') as output:
    for ip_address in re.finditer(ip_address_finder, mbox_file):
        output.write(ip_address.group() + '\n')

```
