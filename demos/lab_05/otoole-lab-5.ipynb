```python
#3 Retrieve JSON data for items

# import requests library, json, and path

import requests

import json

from pathlib import Path

# establish variables

baseURL = 'http://www.loc.gov/'
itemEndpoint = 'item/'
itemIDs = ["98508155", "2017843202", "2020722343"]

# use 'fo' to ask for page data in JSON

parameters = {'fo' : 'json'}

# Make a folder for outputs to live
outputs = Path("json_data_files")
outputs.mkdir(parents=True, exist_ok=True)

# build the request

for itemID in itemIDs:
    r = requests.get(baseURL + itemEndpoint + itemID, params = parameters)
    print(r.url, r.status_code)


#write to json
    data = r.json()

# give a unique name to the file in the folder
    file_path = outputs / f"{itemID}.json"
#load the json file
    with file_path.open("w", encoding="utf-8") as f:
        json.dump(data, f, indent=4)

    print("saved to json_data_files")
```

```python
# 4. Use a different parameter
baseURL = "https://www.loc.gov/search/"

query = input("Enter as search term: ")
response = requests.get(baseURL, params={"q": query, "fo":"json"})

query_data = response.json()
print(json.dumps(query_data["results"], indent=4))
```
